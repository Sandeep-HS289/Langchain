```python
from langchain.messages import SystemMessage,HumanMessage,AIMessage
#sys_msg=SystemMessage("")
msg=[
    SystemMessage("You are a poetry Expert"), #sys_msg,
    HumanMessage("Write a poem on Peace"),
]
response=model.invoke(msg)
response.content
```

```python
### Detailed System msg
from langchain.messages import SystemMessage,HumanMessage
sys_msg=SystemMessage("""You are a Senior developer with ultimate experience and Expert
Always Detail point to point wise Track in Explanations.""")
message=[
    sys_msg,
    HumanMessage("How do i create a AI automation Applications ?")
]
response=model.invoke(message)
print(response.content)
```

we are able pass role,content,MetaData

---> Role -identify's the message type (System,User....) 


---> Content -Represents the actual content of the message(like text,img,audio,docs,etc...)


---> Metadata -Optional fields such as response information,message IDs,and token usage (data about data)

```python
## message metadata
Human_msg=HumanMessage(
    content="hello chat",
    name="alice",
    id="msg_123"

)
```

```python
response=model.invoke([Human_msg])
print(response)
```

```python
from langchain.messages import AIMessage,HumanMessage,SystemMessage

#create an AI message  manually 
ai_msg=AIMessage("I'd be happy to help you!")

msg=[
    SystemMessage("You are a Helpful assistant"),
    HumanMessage("Can you help me ?"),
    ai_msg,
    HumanMessage("That's Sounds Good what is 2+2")
]
response=model.invoke(msg)
print(response.content)
```

```python
response.usage_metadata
```

```python
from langchain.messages import HumanMessage, AIMessage, ToolMessage

ai_msg = AIMessage(
    content="",
    tool_calls=[
        {
            "name": "getweather",
            "args": {"location": "New York"},
            "id": "call_123"
        }
    ]
)

weather_result = "Cool, 21°C"

tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"   # must match the tool call ID
)

messages = [
    HumanMessage(content="What's the weather in New York?"),
    ai_msg,          # AI decided to call the tool
    tool_message     # result returned by the tool
]

response = model.invoke(messages)

print(response.content)
```
