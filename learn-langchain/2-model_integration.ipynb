Streaming : display output while generating output of llm

without Streaming : llm generate total output and then display.

```python
model.invoke("write me a 200 words paragragh on trees.")
```

```python
for chunk in model.stream("Write me a 200-word paragraph."):
    print(chunk.text, end="", flush=True)
```

Batch : if user give multiple inputs --> parallel execution and Gives parallel outputs. Result independent  

```python
responses=model.batch(
    [
        "why parrots talk ?",
        "how humans intelligent?",
        "why maths is so overloved by people ?"
    ]
)
for i in responses:
    print(i)
```

```python
model.batch(
    [
        "why parrots talk ?",
        "how humans intelligent?",
        "why maths is so overloved by people ?"
    ],
    config={
        "max_conncurrency":5, #limit to 5 parallel calls
    }
)
```
